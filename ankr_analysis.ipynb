# Anker Innovations Cross-border E-commerce Sales Analysis
This notebook explores the revenue performance of Anker Innovations across product categories from 2021 to 2024.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Set sketch / notability hand-drawn style
plt.rcParams['font.family'] = ['Arial']
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False

# Dataset: Anker revenue by category (Unit: Billion RMB)
data = {
    'Year': [2021,2021,2021,2021, 2022,2022,2022,2022, 2023,2023,2023,2023, 2024,2024,2024,2024],
    'Category': ['Charging','Audio','Smart Home','Accessories']*4,
    'Revenue': [28.2,12.5,9.3,4.1, 30.5,14.2,11.6,4.8, 33.1,16.8,14.5,5.3, 36.7,19.4,17.2,6.0]
}
df = pd.DataFrame(data)
df.head()
```

```python
# Chart1: Grouped Bar Chart (分组条形图)
pivot_df = df.pivot(index='Year', columns='Category', values='Revenue')
ax1 = pivot_df.plot(kind='bar', figsize=(10,6), width=0.7)
ax1.set_title('Anker Revenue by Product Category (2021–2024)', fontsize=14)
ax1.set_ylabel('Revenue (Billion RMB)')
ax1.set_xlabel('Year')
plt.legend(title='Product Category')
plt.tight_layout()
plt.savefig('grouped_bar_anker.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Chart2: Stacked Bar Chart (堆叠柱状图)
ax2 = pivot_df.plot(kind='bar', stacked=True, figsize=(10,6), width=0.7)
ax2.set_title('Anker Total Revenue Composition (2021–2024)', fontsize=14)
ax2.set_ylabel('Revenue (Billion RMB)')
ax2.set_xlabel('Year')
plt.legend(title='Product Category')
plt.tight_layout()
plt.savefig('stacked_bar_anker.png', dpi=300, bbox_inches='tight')
plt.show()
```

## Analysis Summary
1. Charging products remain Anker’s core revenue source and keep steady growth year by year.
2. Smart home and audio segments have the fastest growth rate, showing successful product diversification.
3. Total revenue grows continuously, which proves Anker’s competitiveness in the global cross-border e-commerce market.
4. Limitation: This simulation dataset is simplified for portfolio demonstration, not full official financial report.
